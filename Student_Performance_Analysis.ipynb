# Student Performance Data Analysis

The main objective is to understand students' marks, study time, attendance, and other factors that may influence academic performance.

- Python
- Jupyter Notebook

## Dataset
Student Performance Dataset – Mathematics (UCI Machine Learning Repository).

## Project Objectives
1. Load and understand the dataset.
2. Perform statistical analysis.
3. Filter and sort student records.
4. Analyse performance using GroupBy.
5. Create at least five meaningful visualizations.
6. Identify important insights from the data.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
df = pd.read_csv("dataset/student-mat.csv", sep=";")

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
print("Number of students:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print("Column Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
df.describe()
```

```python
important_columns = ["sex", "age", "studytime", "absences", "G1", "G2", "G3"]

df[important_columns].head()
```

```python
marks = df["G3"].to_numpy()

print("First 10 Student Marks:", marks[:10])
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
print("Standard Deviation:", np.std(marks))
```

```python
high_performers = df[df["G3"] >= 15]

print("High Performing Students:")
display(high_performers[["sex", "studytime", "G3"]].head(10))

print("Total High Performing Students:", len(high_performers))
```

```python
sorted_students = df.sort_values(by="G3", ascending=False)

print("Top 10 Students:")
display(sorted_students[["sex", "studytime", "G3"]].head(10))
```

```python
study_analysis = df.groupby("studytime")["G3"].mean()

print("Average Marks According to Study Time:")
display(study_analysis)
```

```python
import os

os.makedirs("visualizations", exist_ok=True)

print("Visualizations folder is ready!")
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["G3"], bins=10, kde=True)

plt.title("Distribution of Student Final Marks")
plt.xlabel("Final Marks")
plt.ylabel("Number of Students")

plt.savefig("visualizations/01_marks_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(data=df, x="sex", y="G3", errorbar=None)

plt.title("Average Final Marks by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Marks")

plt.savefig("visualizations/02_gender_performance.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="studytime", y="G3")

plt.title("Study Time vs Student Performance")
plt.xlabel("Study Time Category")
plt.ylabel("Final Marks")

plt.savefig("visualizations/03_studytime_performance.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="G1", y="G3")

plt.title("First Period Marks vs Final Marks")
plt.xlabel("First Period Marks (G1)")
plt.ylabel("Final Marks (G3)")

plt.savefig("visualizations/04_previous_vs_final_marks.png")
plt.show()
```

```python
plt.figure(figsize=(10, 7))

numeric_data = df.select_dtypes(include=np.number)
correlation = numeric_data.corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Student Performance Correlation Heatmap")

plt.savefig("visualizations/05_correlation_heatmap.png")
plt.show()
```

```python
print("KEY INSIGHTS FROM STUDENT PERFORMANCE DATA")
print("-" * 50)

print("1. Total number of students:", len(df))

print("2. Average final marks:", round(df["G3"].mean(), 2))

print("3. Highest final marks:", df["G3"].max())

print("4. Lowest final marks:", df["G3"].min())

print("5. Average marks by gender:")
display(df.groupby("sex")["G3"].mean().round(2))

print("6. Average marks by study time:")
display(df.groupby("studytime")["G3"].mean().round(2))

print("7. Average marks by number of absences:")
display(df.groupby("absences")["G3"].mean().head(10).round(2))

print("8. Correlation between first period and final marks:",
      round(df["G1"].corr(df["G3"]), 2))

print("9. Number of students scoring 15 or above:",
      len(df[df["G3"] >= 15]))

print("10. Average number of absences:",
      round(df["absences"].mean(), 2))
```
